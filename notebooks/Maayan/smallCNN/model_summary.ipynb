# Bearing Fault Diagnosis: Model Summary

## 1. Standardizing the data

I standardized the data using z-score standardization: calculate one overall mean and std from the training data only, then use those same metrics to rescale the training, validation, and test data. I did this because how spread out or amplified each signal is is one of the most predictive features about which fault is present.

## 2. Model choice: CNN

At first I wanted to train a Decision Tree Model, but since each recording is a single long sequence (5,000 time steps), and the distinctive signature of a fault isn't't always at the same spot in every recording, I realized that this would make the Decision Tree model not very capable of predicting the fault accurately. Since CNNs are specifically good at recognizing a pattern no matter where in the sequence it shows up, I decided to train this model instead.

## 3. Results

- **Validation accuracy:** reached 100% by epoch 18 and stayed there through all 50 epochs
- **Test accuracy:** **99.0%** (101 out of 102 test recordings correctly classified)
- Training and validation loss dropped steadily across all 50 epochs.

## 4. Confusion matrix

![Confusion matrix](confusion_matrix.png)

- All 34 'Normal' test recordings were identified correctly. The model reliably tells a healthy bearing apart from a faulty one.
- All 34 'Inner Race Fault' test recordings were identified correctly.
- 33 of 34 'Outer Race Fault' recordings were identified correctly; the one mistake was mislabeled as Inner Race Fault.
- The only error in the whole test set was between the two fault types, never between "healthy" and "faulty." Inner and Outer Race Faults are both bearing-defect signatures and look more similar to each other than either does to a Normal recording, ao this confusion makes sense.
